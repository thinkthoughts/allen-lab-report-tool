## 1. Basic DNA → Residue Encoding

```python
dna_example = "ATGCATGCAGCTAGCTAGCT"

print("Original DNA:", dna_example)
encoded = encode_dna_to_residues(dna_example)
print("Encoded Residues (mod 30):", encoded)
```

## 2. Complementary Pairing as Modular Symmetry

```python
def complement_residue(r: int, mod: int = 30) -> int:
    """Additive inverse for complementary pairing"""
    return (-r) % mod

seq = encode_dna_to_residues("ATGCATGC")
comp = np.array([complement_residue(r) for r in seq])

print("Original residues:   ", seq)
print("Complementary:       ", comp)
print("Sum (mod 30):        ", (seq + comp) % 30)
```

## 3. Simulated Chromosome 21 Segment

```python
chr21_segment = "ATGCGATCGATCGATCGTAGCTAGCTAGCT"
encoded_chr21 = encode_dna_to_residues(chr21_segment)

print(f"Encoded segment length: {len(encoded_chr21)}")
print("First 25 residues:", encoded_chr21[:25])
```

## 4. CGCS Comparison: Normal vs Trisomy

```python
print("CGCS Impact of Extra Chromosome:\n")
for ratio, label in [(1.0, "Normal"), (1.5, "Trisomy 21")]:
    score = trisomy_cgcs_score(dosage_ratio=ratio)['cgcs']
    print(f"{label:12} → CGCS = {score:.4f}")
```

## 5. Residue Distribution

```python
plt.figure(figsize=(10, 5))
plt.hist(encoded_chr21, bins=range(31), alpha=0.75, color='steelblue', edgecolor='black')
plt.title("Residue Distribution in Simulated Chromosome 21 Segment")
plt.xlabel("Residue Value (mod 30)")
plt.ylabel("Frequency")
plt.grid(True, alpha=0.3)
plt.show()
```
