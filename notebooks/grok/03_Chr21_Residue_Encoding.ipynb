```python
# ================================================
# SETUP
# ================================================
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd() / "src"))

import grok
from grok.residue_encoding import *
from grok.trisomy_metrics import trisomy_cgcs_score

import numpy as np
import matplotlib.pyplot as plt

print("✅ Notebook 03 Ready")
print(f"Version: {grok.__version__}")
```

## 1. Basic DNA Base Encoding

```python
dna_example = "ATGCATGCAGCTAGCT"

print("DNA Sequence:", dna_example)
encoded = encode_dna_to_residues(dna_example)
print("Encoded Residues (mod 30):", encoded)
```

## 2. Complementary Pairing as Modular Symmetry

```python
def complement_residue(r: int, mod: int = 30) -> int:
    return (-r) % mod

residues = encode_dna_to_residues("ATGCATGC")
complements = np.array([complement_residue(r) for r in residues])

print("Original:    ", residues)
print("Complements: ", complements)
print("Sum mod 30:  ", (residues + complements) % 30)
```

## 3. Simulated Chromosome 21 Segment

```python
chr21_short = "ATGCGATCGATCGATCGTAGCTAGCTAGCT"
encoded_chr21 = encode_dna_to_residues(chr21_short)

print(f"Length of encoded segment: {len(encoded_chr21)}")
print("First 30 residues:", encoded_chr21[:30])
```

## 4. CGCS Impact of Duplication

```python
print("CGCS Comparison:\n")

for ratio, name in [(1.0, "Normal"), (1.5, "Trisomy 21")]:
    score = trisomy_cgcs_score(dosage_ratio=ratio)['cgcs']
    print(f"{name:12} → CGCS = {score:.4f}")
```

## 5. Residue Distribution Visualization

```python
plt.figure(figsize=(10, 5))
plt.hist(encoded_chr21, bins=range(31), alpha=0.75, color='steelblue', edgecolor='black')
plt.title("Residue Distribution in Simulated Chromosome 21 Segment")
plt.xlabel("Residue Value (mod 30)")
plt.ylabel("Count")
plt.grid(True, alpha=0.3)
plt.show()
```
