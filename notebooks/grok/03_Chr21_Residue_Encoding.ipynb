## 1. Basic DNA Base → Residue Mapping

```python
# Example DNA sequence
dna_seq = "ATGCATGCAGCTAGCTAG"

print("Original DNA:", dna_seq)
print("Encoded residues (mod 30):", encode_dna_to_residues(dna_seq))

for base in "ATGC":
    print(f"{base} → {encode_base(base)}")
```

## 2. Complementary Base-Pairing as Modular Symmetry

```python
def complement_residue(res: int, mod: int = 30) -> int:
    """Simple complement: additive inverse mod N"""
    return (-res) % mod

seq = encode_dna_to_residues("ATGCATGC")
comp = np.array([complement_residue(r) for r in seq])

print("Original residues:", seq)
print("Complementary residues:", comp)
print("Sum (should be 0 mod 30):", (seq + comp) % 30)
```

## 3. Chromosome 21 Segment Simulation

```python
# Simulate a short chr21 segment
chr21_segment = "ATGCGATCGATCGATCGTAGCTAGCT"
encoded = encode_dna_to_residues(chr21_segment)

print(f"Encoded length: {len(encoded)}")
print("First 20 residues:", encoded[:20])

# CGCS on normal vs duplicated segment (simplified)
normal_cgcs = trisomy_cgcs_score(dosage_ratio=1.0)['cgcs']
duplicated_cgcs = trisomy_cgcs_score(dosage_ratio=1.5)['cgcs']

print(f"\nNormal dosage CGCS: {normal_cgcs:.4f}")
print(f"Duplicated (Trisomy-like) CGCS: {duplicated_cgcs:.4f}")
```

## 4. Visualizing Residue Distribution

```python
plt.figure(figsize=(10, 4))
plt.hist(encoded, bins=30, alpha=0.7, color='blue', edgecolor='black')
plt.title("Residue Distribution in Simulated Chromosome 21 Segment")
plt.xlabel("Residue Value (mod 30)")
plt.ylabel("Frequency")
plt.grid(True, alpha=0.3)
plt.show()
```

**Key Takeaway**:  
We can represent DNA base-pairing rules as **symmetry operations on a modular manifold**. Extra chromosome material (Trisomy 21) introduces redundancy and dosage imbalance that lowers CGCS.
